```python

from openai import OpenAI
import os

client = OpenAI(
    api_key=os.environ["GROQ_API_KEY"],
    base_url="https://api.groq.com/openai/v1"
)

WORKING_MODEL = None

for model in ["openai/gpt-oss-120b", "qwen/qwen3.8-27b"]:
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[
                {"role": "user", "content": "Reply only with OK"}
            ],
            max_tokens=20
        )
        WORKING_MODEL = model
        print("SUCCESS:", model)
        print(response.choices[0].message.content)
        break
    except Exception as e:
        print("FAILED:", model, "-", str(e)[:250])

if WORKING_MODEL is None:
    raise RuntimeError(
        "Neither model worked. Check the error messages above."
    )
```

```python
import os
from crewai import Agent, Task, Crew, Process, LLM

# ==========================================================
# 1. CHECK API KEY
# ==========================================================
api_key = os.environ.get("GROQ_API_KEY")

if not api_key:
    raise ValueError(
        "Groq API key missing. Run Step 2 first."
    )

print("=" * 65)
print("       CREWAI MULTI-AGENT AI SYSTEM")
print("=" * 65)

# ==========================================================
# 2. CREATE ONE LLM SHARED BY ALL AGENTS
# ==========================================================


# ==========================================================
# 2. CREATE ONE LLM SHARED BY ALL AGENTS
# ==========================================================


llm = LLM(
    model=f"openai/{WORKING_MODEL}",
    base_url="https://api.groq.com/openai/v1",
    api_key=api_key,
    temperature=0.2
)

print("Working model configured:", WORKING_MODEL)

print("Groq GPT-OSS 120B configured successfully!")
print("One shared LLM will be used by all three agents.")
print("Groq LLM configured successfully!")
print("\nOne Groq LLM configured successfully.")

# ==========================================================
# 3. CREATE AGENT 1: RESEARCHER
# ==========================================================
researcher = Agent(
    role="AI Researcher",
    goal="Collect and organize useful information about {topic}",
    backstory=(
        "You are a careful researcher who explains topics "
        "clearly and identifies important facts, applications, "
        "benefits, and limitations."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False
)

# ==========================================================
# 4. CREATE AGENT 2: ANALYST
# ==========================================================
analyst = Agent(
    role="Research Analyst",
    goal="Analyze the research about {topic} and identify insights",
    backstory=(
        "You are an analytical thinker. You examine research, "
        "identify important patterns, discuss benefits and risks, "
        "and produce balanced conclusions."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False
)

# ==========================================================
# 5. CREATE AGENT 3: REPORT WRITER
# ==========================================================
writer = Agent(
    role="Report Writer",
    goal="Prepare a clear and well-structured report about {topic}",
    backstory=(
        "You are a professional report writer who turns research "
        "and analysis into an understandable report with headings, "
        "examples, conclusions, and recommendations."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False
)

# ==========================================================
# 6. DEFINE TASK 1: RESEARCH
# ==========================================================
research_task = Task(
    description=(
        "Research the topic: {topic}.\n"
        "Explain its meaning, key concepts, applications, "
        "benefits, challenges, and practical examples.\n"
        "Organize the information into clear sections. "
        "Do not invent statistics or claim to have verified "
        "live information."
    ),
    expected_output=(
        "A structured research summary covering definitions, "
        "key concepts, applications, benefits, and challenges."
    ),
    agent=researcher
)

# ==========================================================
# 7. DEFINE TASK 2: ANALYSIS
# ==========================================================
analysis_task = Task(
    description=(
        "Analyze the research provided by the previous task "
        "about {topic}.\n"
        "Identify the most important insights, advantages, "
        "limitations, possible risks, and future opportunities. "
        "Use the research as your basis and avoid unsupported claims."
    ),
    expected_output=(
        "An analytical summary with key insights, balanced "
        "advantages and disadvantages, and practical conclusions."
    ),
    agent=analyst,
    context=[research_task]
)

# ==========================================================
# 8. DEFINE TASK 3: FINAL REPORT
# ==========================================================
report_task = Task(
    description=(
        "Write the final report about {topic}, using the "
        "research and analysis from the previous tasks.\n"
        "Include:\n"
        "1. Title\n"
        "2. Introduction\n"
        "3. Key concepts and applications\n"
        "4. Benefits\n"
        "5. Challenges and risks\n"
        "6. Future scope\n"
        "7. Conclusion\n"
        "Keep the report clear, informative, and suitable "
        "for a student presentation."
    ),
    expected_output=(
        "A complete, well-organized report in Markdown "
        "with all seven requested sections."
    ),
    agent=writer,
    context=[research_task, analysis_task]
)

# ==========================================================
# 9. CREATE THE CREW
# ==========================================================
crew = Crew(
    agents=[researcher, analyst, writer],
    tasks=[research_task, analysis_task, report_task],
    process=Process.sequential,
    verbose=True
)

# ==========================================================
# 10. RUN THE CREW
# ==========================================================
topic = input(
    "\nEnter a topic for the AI crew "
    "(example: Artificial Intelligence in Healthcare): "
).strip()

if not topic:
    topic = "Artificial Intelligence in Healthcare"

print("\n" + "=" * 65)
print("STARTING CREW EXECUTION")
print("Topic:", topic)
print("LLM: One shared Groq model")
print("Agents: Researcher -> Analyst -> Writer")
print("=" * 65)

result = await crew.kickoff_async(inputs={"topic": topic})

# ==========================================================
# 11. DISPLAY FINAL OUTPUT
# ==========================================================
print("\n\n" + "=" * 65)
print("                 FINAL RESEARCH REPORT")
print("=" * 65)
print(result)
print("\n" + "=" * 65)
print("CREW EXECUTION COMPLETED")
print("=" * 65)
```
